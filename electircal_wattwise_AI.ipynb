```python
import pandas as pd
import numpy as np

df = pd.read_csv("electricity_consumption_based_weather_dataset.csv")
df.head(4)
```

```python
print(df.size)
print(df.shape)
```

```python
df.info()
```

```python
#converting date column from string/ object to date
df["date"] = pd.to_datetime(df['date'], errors='coerce')
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
for col in df.columns:
  if df[col].isnull().sum() > 0 and col != "date":
    n = df[col].isnull().sum()
    df[col] = df[col].interpolate(method= "linear")
    print(f"filled {n} missing values in {col} column")
df.info()
```

```python
print(f"we have {df.duplicated().sum()} duplicates")
print(f"we have {df['date'].duplicated().sum()} duplicated dates")
```

```python
df_sorted = df.sort_values(by = 'date')
print(df_sorted.head(4))
df.head(4)
```

```python
full_range = pd.date_range(df_sorted['date'].min(),(df_sorted['date']).max())
missing_dates = full_range.difference(df_sorted['date'])
print(missing_dates)
```

```python
wrong_temp = df_sorted[df_sorted['TMIN']>df_sorted['TMAX']]
print(f"we have {len(wrong_temp)} wrong recorded tempratures")

for col in ["daily_consumption","PRCP","AWND"]:
  print(f"we have {(df_sorted[col]<0).sum()} nergative values in {col} columns")

q1 = df_sorted['daily_consumption'].quantile(0.25)
q3 = df_sorted['daily_consumption'].quantile(0.75)
iqr = q3-q1
lower_bound = q1 - 1.5*iqr
upper_bound = q3 + 1.5*iqr
outliers = df_sorted[(df_sorted['daily_consumption']<lower_bound) | (df_sorted['daily_consumption']>upper_bound)]

print(outliers.shape)
print(outliers.head(4))
```

```python
df_sorted = df.set_index('date').reindex(full_range).rename_axis('date').reset_index()
df_sorted = df_sorted.interpolate(method='linear')

print(df_sorted.shape)
df_sorted.head(4)
```

```python
print("number of duplicated rowa is: ",df_sorted.duplicated().sum())
print()
print(df_sorted.info())
```

```python
import matplotlib.pyplot as plt

fig, axis = plt.subplots(2,2,figsize=(20,10))

axis[0,0].plot(df_sorted['date'],df_sorted['daily_consumption'])
axis[0,0].set_title("daily consumption over time")

axis[0,1].hist(df_sorted['daily_consumption'],bins = 30)
axis[0,1].set_title("distribution of daily consumpution")

axis[1,0].scatter(df_sorted['TMAX'],df_sorted['daily_consumption'])
axis[1,0].set_title("daily consumption vs max temp")

axis[1,1].boxplot(df_sorted['daily_consumption'].dropna())
axis[1,1].set_title("boxplot of daily consumption")

plt.show()
```

```python
df = df_sorted
df['day_of_week'] = df['date'].dt.dayofweek
df['month'] = df['date'].dt.month
df["is_weekend"] = df["day_of_week"].isin([5,6]).astype(int)

print(df[['date','day_of_week','month','is_weekend']].head(4))
```

```python
df['consumpution_lag_1'] = df['daily_consumption'].shift(1)
df['consumpution_lag_7'] = df['daily_consumption'].shift(7)
df[["date","daily_consumption","consumpution_lag_1","consumpution_lag_7"]].head(10)
```

```python
df['rolling_mean_7'] = df['daily_consumption'].rolling(window=7).mean()
df['rolling_max_7'] = df['daily_consumption'].rolling(window=7).max()

df['temp_range'] = (df['TMAX'] - df['TMIN'])
df[['date','temp_range','daily_consumption','rolling_mean_7','rolling_max_7']].head(10)
```

```python
before = df.shape[0]
df = df.dropna().reset_index(drop = True)
after = df.shape[0]
print(f"we have {before-after} rows with missing values")
df.shape
```

```python
splite = int(len(df)*0.8)
training = df.iloc[:splite]
testing = df.iloc[splite:]

print(f" training start from {training['date'].min()} and ends at {training['date'].max()} with {len(training)} rows")
print(f" testing start from {testing['date'].min()} and ends at {testing['date'].max()} with {len(testing)} rows")
```

```python
features_col = ['AWND', 'PRCP', 'TMAX', 'TMIN','day_of_week', 'month',
                'is_weekend', 'consumpution_lag_1',
                'consumpution_lag_7', 'rolling_mean_7', 'rolling_max_7']
```

```python
X_train = training[features_col]
y_train = training['daily_consumption']

X_test = testing[features_col]
y_test = testing['daily_consumption']
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,mean_absolute_error, r2_score

lr = LinearRegression()
lr.fit(X_train,y_train)

y_pred = lr.predict(X_test)

mae_lr = mean_absolute_error(y_test,y_pred)
rmse_lr = np.sqrt(mean_squared_error(y_test,y_pred))
mape_lr = np.mean(np.abs((y_test - y_pred) / y_test)) * 100
r2_lr = r2_score(y_test,y_pred)

print('Lenier Regression Metrics:')
print(f'Mean Absolute Error (MAE): {mae_lr:.2f}')
print(f'Root Mean Squared Error (RMSE): {rmse_lr:.2f}')
print(f'Mean Absolute Percentage Error (MAPE): {mape_lr:.2f}%')
```

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(n_estimators=200, random_state=42)
rf.fit(X_train, y_train)

y_pred = rf.predict(X_test)
mae_rf = mean_absolute_error(y_test, y_pred)
rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred))
mape_rf = np.mean(np.abs((y_test - y_pred) / y_test)) * 100
r2_rf = r2_score(y_test, y_pred)

print('Random Forest Metrics:')
print(f'Mean Absolute Error (MAE): {mae_rf:.2f}')
print(f'Root Mean Squared Error (RMSE): {rmse_rf:.2f}')
print(f'Mean Absolute Percentage Error (MAPE): {mape_rf:.2f}%')
```

```python
feature_impotance = pd.DataFrame({'feature': features_col, 'importance': rf.feature_importances_})
feature_impotance = feature_impotance.sort_values('importance', ascending=False)
print(feature_impotance)
```
